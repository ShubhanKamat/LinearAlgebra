Write an algorithm that computes the norm of a vector 

```python
import numpy as np

def cal_norm(vector):

    norm = 0

    for i in vector:
        
        norm += i**2

    return norm**(1/2)

def np_norm(vector):
    return np.sqrt(np.sum(vector**2))

vec1 = np.array([1,2,3,4])

print(cal_norm(vec1))
print(np_norm(vec1))
print(np.linalg.norm(vec1))
```
