Write an algorithm that computes the norm of a vector 

```python
import numpy as np

def cal_norm(vector):

    norm = 0

    for i in vector:
        
        norm += i**2

    return norm**(1/2)

def np_norm(vector):
    return np.sqrt(np.sum(vector**2))

vec1 = np.array([1,2,3,4])

print(cal_norm(vec1))
print(np_norm(vec1))
print(np.linalg.norm(vec1))
```

Create a Python function that will take a vector as input and output a unit vector in
the same direction.

```python
def create_unit_vec(vector):
    norm = np_norm(vector)

    if norm != 0:
        return vector/norm
    else:
        return -1
```
